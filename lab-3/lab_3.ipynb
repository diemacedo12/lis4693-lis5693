Installing libraries

```python
import pandas as pd
import altair as alt
from vega_datasets import data as vega_data
```

Loading the Seattle Circulation data from GitHub

```python
import requests
import io

url = "https://raw.githubusercontent.com/manika-lamba/SP26-LIS4_5693/refs/heads/main/lab-assignments/lab-3/Seattle-Library_2015-2021.csv"
response = requests.get(url)
response.raise_for_status() # Raise an exception for HTTP errors
text = response.text
```

Saving data as a DataFrame

```python
seattle_df = pd.read_csv(io.StringIO(text))
seattle_df.head()
```

Grouping the data by MaterialType and CheckoutYear

```python
grouped = seattle_df.groupby(['MaterialType', 'CheckoutYear']).size().reset_index(name='counts')
```

Visualizing data as an area chart

```python
alt.Chart(grouped).mark_area(opacity=0.7).encode(
    x=alt.X('MaterialType', sort='-y'),
    y='counts',
    color='CheckoutYear:N',
    tooltip=['MaterialType', 'counts', 'CheckoutYear']
).properties(
    width=500,
    height=350
).interactive()
```

Visualizing data as an line chart

```python
alt.Chart(grouped).mark_line(point=True).encode(
    x=alt.X('MaterialType', sort='-y'),
    y='counts',
    color='CheckoutYear:N',
    tooltip=['MaterialType', 'counts', 'CheckoutYear']
).properties(
    width=500,
    height=350
).interactive()
```

Horizontal concatenation using alt.hconcat()

```python
alt.hconcat(
    alt.Chart(grouped).mark_area(opacity=0.7).encode(
        x=alt.X('MaterialType', sort='-y'),
        y='counts',
        color='CheckoutYear:N',
        tooltip=['MaterialType', 'counts', 'CheckoutYear']
    ).properties(
        width=480,
        height=350,
        title='Area View: Checkouts by Material Type & Year'
    ).interactive(),

    alt.Chart(grouped).mark_line(point=True).encode(
        x=alt.X('MaterialType', sort='-y'),
        y='counts',
        color='CheckoutYear:N',
        tooltip=['MaterialType', 'counts', 'CheckoutYear']
    ).properties(
        width=620,
        height=350,
        title='Line View: Checkouts by Material Type & Year'
    ).interactive()
)
```
