```python
#Imports
import spacy
import requests
from spacy.matcher import PhraseMatcher
from collections import Counter
```

Loading transcript from GitHub repo directly

```python
#Loading and reading file
url = "https://raw.githubusercontent.com/diemacedo12/lis4693-lis5693/main/transcript.txt"
response = requests.get(url)
response.raise_for_status()
text = response.text
```

Number of characters + first 100 characters

```python

print("Number of characters:", len(text))
print(text[:100])
```

Sentence segmentation using blank pipeline

```python
#Using a blank pipeline
nlp_blank = spacy.blank("en")
nlp_blank.add_pipe("sentencizer")

doc_blank = nlp_blank(text)
```

```python
#Sentence segmentation
sentences = list(doc_blank.sents)

print("Number of sentences:", len(sentences))
print("\nFirst 3 sentences:")
print(sentences[0])
print(sentences[1])
print(sentences[2])
```

Word count and unique words

```python
#Word count and unique words
words = [token.text.lower() for token in doc_blank if token.is_alpha]

print("Total number of words:", len(words))
print("Number of unique words:", len(set(words)))
```

Most frequent words

```python
#Most frequent word
word_freq = Counter(words)

print("Most frequent word:")
print(word_freq.most_common(1))
```

The word "the" is one of the most common words in the English language.

```python
#Using Full spaCy Pipeline
nlp2 = spacy.load("en_core_web_sm")
doc2 = nlp2(text)

print("Number of named entities found:", len(doc2.ents))

print("\nEntity types that appear:")
types = set()
for ent in doc2.ents:
    types.add(ent.label_)
print(", ".join(sorted(types)) if types else "None")

print("\nSome named entities:")
for ent in list(doc2.ents)[:10]:
    print(ent.text, ent.label_)
```

```python
#PhraseMatcher
matcher = PhraseMatcher(nlp2.vocab, attr="LOWER")

phrases = ["Google search results", "organic results", "SerpApi"]
patterns = [nlp2(p) for p in phrases]

matcher.add("RELEVANT_PHRASES", patterns)

matches = matcher(doc2)

print("Matches found:")
for match_id, start, end in matches:
    print(doc2[start:end])
    print("Sentence:", doc2[start].sent)
```
